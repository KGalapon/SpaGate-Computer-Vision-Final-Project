```python
import numpy as np
from datasets.dataset_class import read_annotation_file

names = ["person", "car", "bus", "bicycle", "motorcycle"]
for split in ["test_clean", "test_foggy", "rtts"]:
    boxes = np.concatenate([b for _, b in read_annotation_file(split)])
    print(split)
    for c, name in enumerate(names):
        bc = boxes[boxes[:, 4] == c]
        side = np.sqrt((bc[:, 2] - bc[:, 0]) * (bc[:, 3] - bc[:, 1])) if len(bc) else np.array([0])
        print(f"  {name:10s} n={len(bc):5d}  median side {np.median(side):.0f}px")
```
